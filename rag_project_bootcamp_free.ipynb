# loading the  file

```python
from langchain_community.document_loaders import TextLoader

```

```python
doc=TextLoader("/content/Saravanan.txt")
document_text=doc.load()
```

```python
document_text
```

# text splitter

```python
!pip install -qU langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=2
)

chunks = splitter.split_documents(document_text)
```

```python
chunks
```

# embedding ...

```python
pip install langchain_google_genai

```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview", api_key="REMOVED_API_KEY")
```

# vector db - pip install faiss-cpu

```python
pip install faiss-cpu
```

```python
from langchain_community.vectorstores import FAISS

vector_db= FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)


```

```python
retrival =vector_db.as_retriever(search_kwargs={"k":5})
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt_tem=ChatPromptTemplate([
    """

your are a helpfull assistant , answer the questions using the context ,
dont explain everything just give me in 2 to 3 lines summary
context:{context}
question:{question}
    """

])
```

```python
!pip install -qU langchain-google-genai
from langchain_core.runnables import RunnablePassthrough
```

```python
from langchain_google_genai import GoogleGenerativeAI
client = GoogleGenerativeAI(model="gemini-2.5-flash", google_api_key="REMOVED_API_KEY")
```

```python
chain  = ({
    "context":retrival,
    "question":RunnablePassthrough()
})  | prompt_tem  | client
```

```python
res=chain.invoke("what is the debut film for surya ?")
```
